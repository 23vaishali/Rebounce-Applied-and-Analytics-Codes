**Problem 5. The Password Strength check**

```python
password = "hello@2026"
lenght= len (password) >=8
symbol_contain = "@" in password
strenght = lenght and symbol_contain
print(f"Password:{password}")
print(f"Lenght OK (>= 8): {lenght}")
print(f"Contain '@': {symbol_contain}")
print(f"Strong : {strenght}")
```
